## Local versus remote transport
Stdio starts a subprocess. Streamable HTTP connects to a hosted endpoint. This optional example only lists the remote tools.

```python
from agents.mcp import MCPServerStreamableHttp
async with MCPServerStreamableHttp(name="Context7", params={"url": "https://mcp.context7.com/mcp", "timeout": 60}) as server:
    tools = await server.list_tools()
    print([tool.name for tool in tools])
```
